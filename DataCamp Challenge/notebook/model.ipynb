```python
df['is_et'] = df['Win conditions'].str.strip().ne('')
```

```python
def clean_data(txt: str):
    sep = txt.split('(')[1].rstrip(' ) ')
    clean_txt = sep.split(' - ')
    return clean_txt
```

```python
for i, row in df.iterrows():
    if row['is_et']:
        # print(i)
        # print(row['Win conditions']) debug purpose
        if 'extra time' not in row['Win conditions'] and 'Golden Goal' not in row['Win conditions']:
            goals = clean_data(row['Win conditions'])
            home_goal = goals[0]
            away_goal = goals[1]
            # print(home_goal)
            df.loc[i, "Home Team Goals"] = home_goal
            df.loc[i, "Away Team Goals"] = away_goal
    else:
        # print("Row Skipped")
        pass
```

```python
df['Winner'] = df.apply(lambda row: row['Home Team Name'] if row['Home Team Goals'] > row['Away Team Goals'] else (row['Away Team Name'] if row['Home Team Goals'] < row['Away Team Goals'] else 'draw'), axis=1)
```

```python
team_list = {}
```

```python
df['Home_Team_Elo'] = 0
df['Away_Team_Elo'] = 0
```

```python
def elo_rating(team):
    if team not in team_list:
        team_list[team] = 1500
        return team_list[team]
    else:
        return team_list[team]
```

```python
# Functions for elo rating calculation
Home_Adv = 1.10
def calculate_elo_rating(team_a, team_b, stage, winner, Win_Con):

    # print(type(team_a)) For debug

    # Get current ratings
    rating_a = elo_rating(team_a)
    rating_b = elo_rating(team_b)

    # Calculate expected scores
    expected_a = 1 / (1 + 10 ** ((rating_b - rating_a) / 400))
    expected_b = 1 / (1 + 10 ** ((rating_a - rating_b) / 400))

    # Determine actual scores based on winner
    if winner == team_a:
        score_a, score_b = 1, 0
    elif winner == team_b:
        score_a, score_b = 0, 1
    else:  # Draw
        score_a, score_b = 0.5, 0.5

    if stage in ['Group A', 'Group B', 'Group C', 'Group D', 'Group E', 'Group F', 'Group G', 'Group H']:
        k_factor = 20  # Lower K-factor for group stage
        rating_a += k_factor * (score_a - expected_a) * Home_Adv
        rating_b += k_factor * (score_b - expected_b)

        team_list[team_a] = rating_a
        team_list[team_b] = rating_b

        return rating_a, rating_b
    
    elif stage in ['Round of 16', 'Quarter-finals', 'Semi-finals', 'Final', 'Match for third place', 'Play-off for third place', 'Third place']:
        k_factor = 40  # Higher K-factor for knockout stage
        rating_a += k_factor * (score_a - expected_a)
        rating_b += k_factor * (score_b - expected_b)

        team_list[team_a] = rating_a
        team_list[team_b] = rating_b

        return round(rating_a, 2), round(rating_b,2)
```

```python
for i, row in df.iterrows():
    Home_Team = row['Home Team Name']
    Away_Team = row['Away Team Name']
    Stage = row['Stage']
    Winner = row['Winner']
    Winner_Condition = row['Win conditions']

    if i == 642:
        print(Home_Team, Away_Team, Stage, Winner)
    home_value, away_value = calculate_elo_rating(Home_Team, Away_Team, Stage, Winner, Winner_Condition)

    df.loc[i, "Home_Team_Elo"] = home_value
    df.loc[i, "Away_Team_Elo"] = away_value

    # print(f"Updated row {i}")
    
```
